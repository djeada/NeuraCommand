# Perceptron Learning (Linear Model)

Train a single neuron to learn a linear relationship and track weight updates.


```python
import numpy as np
import matplotlib.pyplot as plt
import sys

sys.path.insert(0, '../src')

from neura_command import Sequential, Dense, SGD, MSELoss

np.random.seed(42)
```

## 1. Dataset


```python
rng = np.random.default_rng(42)
X = np.linspace(-1, 1, 20).reshape(-1, 1)
y = 2 * X + rng.normal(0, 0.2, size=X.shape)
```

## 2. Model Setup


```python
model = Sequential(
    layers=[Dense(1, 1)],
    optimizer=SGD(learning_rate=0.2),
    loss=MSELoss(),
)

initial_line_x = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
initial_line_y = model.predict(initial_line_x)
```

## 3. Train and Track Weights


```python
epochs = 20
training_history = []

model.train_mode()
for epoch in range(epochs):
    preds = model.forward(X, training=True)
    loss = model.loss.compute(y, preds)
    grad = model.loss.derivative(y, preds)
    model.backward(grad)
    model.optimizer.step(model.layers)

    weight = float(model.layers[0].weights[0, 0])
    bias = float(model.layers[0].bias[0, 0])
    training_history.append({
        'epoch': epoch + 1,
        'weight': weight,
        'bias': bias,
        'loss': float(loss),
    })
```

## 4. Fit Before vs. After


```python
trained_line_y = model.predict(initial_line_x)

plt.figure(figsize=(10, 4))
plt.scatter(X, y, label='Data', alpha=0.6)
plt.plot(initial_line_x, initial_line_y, linestyle='--', color='gray', label='Initial fit')
plt.plot(initial_line_x, trained_line_y, color='red', label='Trained fit')
plt.title('Perceptron Fit')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
