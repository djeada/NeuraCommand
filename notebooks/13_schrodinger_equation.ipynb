# Schrodinger Equation: Learning a Wavefunction

Approximate the ground-state wavefunction of a 1D infinite square well with a small MLP.


```python
import numpy as np
import matplotlib.pyplot as plt
import sys

sys.path.insert(0, '../src')

from neura_command import Sequential, Dense, Tanh, Adam, MSELoss

np.random.seed(42)
```

## 1. Generate Training Data


```python
x = np.linspace(0, 1, 200).reshape(-1, 1)
psi = np.sqrt(2) * np.sin(np.pi * x)

indices = np.random.permutation(len(x))
train_idx = indices[:160]
test_idx = indices[160:]

X_train, y_train = x[train_idx], psi[train_idx]
X_test, y_test = x[test_idx], psi[test_idx]
```

## 2. Build and Train the Network


```python
model = Sequential(
    layers=[
        Dense(1, 64),
        Tanh(),
        Dense(64, 64),
        Tanh(),
        Dense(64, 1),
    ],
    optimizer=Adam(learning_rate=0.01),
    loss=MSELoss(),
)

history = model.train(X_train, y_train, epochs=500, batch_size=32, verbose=False)
```

## 3. Evaluate


```python
pred_test = model.predict(X_test)

test_mse = np.mean((pred_test - y_test) ** 2)
print(f'Test MSE: {test_mse:.6f}')
```

## 4. Plot Predictions


```python
pred_full = model.predict(x)

plt.figure(figsize=(10, 4))
plt.plot(x, psi, label='Analytic solution')
plt.plot(x, pred_full, linestyle='--', label='NeuraCommand prediction')
plt.title('Ground-State Wavefunction')
plt.xlabel('x')
plt.ylabel('psi(x)')
plt.legend()
plt.show()
```

## 5. Training Loss


```python
plt.figure(figsize=(6, 4))
plt.plot(history.get('loss', []))
plt.title('Training Loss')
plt.xlabel('Epoch')
plt.ylabel('MSE')
plt.show()
```
